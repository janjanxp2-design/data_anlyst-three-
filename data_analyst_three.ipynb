```python
import numpy as np
import pandas as pd

np.random.seed(42)
n_rows = 600

cities = ["Karachi", "Lahore", "Faisalabad", "Rawalpindi", "Multan", "Sialkot", "Peshawar", "Gujranwala"]
departments = ["Production", "Logistics", "Quality Control", "Maintenance", "Packaging", "Inventory"]
shifts = ["Morning", "Evening", "Night"]
job_titles = ["Machine Operator", "Truck Driver", "QC Inspector", "Technician", "Senior Supervisor", "Packer", "Store Keeper", "Electrician"]

first_names = ["Ali", "Usman", "Saima", "Bilal", "Tariq", "Hamza", "Zubair", "Asad", "Kashif", "Imran", "Zahid", "Nabeel", "Fahad", "Omer"]
last_names = ["Khan", "Ahmed", "Riaz", "Hassan", "Mahmood", "Raza", "Sheikh", "Malik", "Jamal", "Ashraf", "Shah", "Ibal"]

data = {
    # 6 String Columns
    "Employee_ID": [f"EMP{i:03d}" for i in range(101, 101 + n_rows)],
    "Employee_Name": [f"{np.random.choice(first_names)} {np.random.choice(last_names)}" for _ in range(n_rows)],
    "City": np.random.choice(cities, n_rows),
    "Department": np.random.choice(departments, n_rows),
    "Shift": np.random.choice(shifts, n_rows),
    "Job_Title": np.random.choice(job_titles, n_rows),
    
    # 14 Integer Columns
    "Age": np.random.randint(20, 58, n_rows),
    "Base_Salary": np.random.randint(35000, 90000, n_rows),
    "Bonus": np.random.randint(2000, 12000, n_rows),
    "Overtime_Hours": np.random.randint(0, 35, n_rows),
    "Food_Expense": np.random.randint(4500, 9500, n_rows),
    "Performance_Score": np.random.randint(60, 100, n_rows),
    "Monthly_Sales_Target": np.random.randint(50, 160, n_rows),
    "Raw_Material_In_Tons": np.random.randint(70, 220, n_rows),
    "Finished_Goods_Out_Tons": np.random.randint(60, 200, n_rows),
    "Gross_Revenue_PKR": np.random.randint(600000, 2500000, n_rows),
    "Electricity_Bill_PKR": np.random.randint(25000, 75000, n_rows),
    "Transport_Cost_PKR": np.random.randint(8000, 30000, n_rows),
    "Net_Profit_PKR": np.random.randint(100000, 700000, n_rows),
    "Tax_Deducted_PKR": np.random.randint(5000, 25000, n_rows)
}

df = pd.DataFrame(data)
df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns 
sns.jointplot(data=df,x='Base_Salary',y='Bonus', kind="reg",height=6)
plt.suptitle('Joinplot of base_salary and Bonus',y=1.02)
plt.show()
```

```python
sns.boxenplot(data=df,x="Food_Expense",hue="City")
plt.grid()
plt.legend(bbox_to_anchor=(1.05,1),loc="upper left")
plt.suptitle("Density of Food expense by city")
plt.show()
```

```python
City=df['City'].value_counts()
plt.pie(City,labels=City.index,autopct="%1.1f%%",wedgeprops={"edgecolor":"white"})
plt.suptitle("Worker viwe countries")
plt.show()
```

```python
sns.histplot(data=df,x="Age",bins=15,kde=True,color="green")
plt.grid()
plt.title("AGE of worker")
plt.show()
```

```python
plt.bar(df["Performance_Score"],df["City"])
plt.title("Performance Score Cityvise")
plt.ylabel("CITIES")
plt.xlabel("SCORE")
plt.show()
```

```python
import seaborn as sns
sns.scatterplot(data=df,x="Performance_Score",y="Monthly_Sales_Target",hue="Department",style="Shift")
plt.title("Scatter data of Department Monthley_salevise hue deparment")
plt.legend(bbox_to_anchor=(1.05,1), loc='upper left')
plt.grid()
plt.show()
```

```python
df['Age'] = pd.cut(
    df['Age'], 
    bins=[18, 30, 40, 50, 70], 
    labels=['18-30', '31-40', '41-50', '51-70']
)
sns.barplot(data=df,x="Age",y="Overtime_Hours",hue="Shift")
plt.title("Gaussing the extent of greed")
plt.legend(bbox_to_anchor=(1.05,1), loc='upper left')
plt.grid()
plt.show()
```

```python
sns.kdeplot(data=df,x='Net_Profit_PKR',hue="Shift",fill=True)
plt.grid()
plt.title('Net profit by City')
plt.show()
```

```python
numeric_data=df[["Age",'Base_Salary','Bonus','Overtime_Hours','Food_Expense','Shift']]
sns.pairplot(data=numeric_data,hue="Shift")
plt.grid()
plt.title("pair plots of our numeric data")
plt.show()
```

```python
job_titles=df['Job_Title'].value_counts()
plt.pie(job_titles,labels=job_titles.index,autopct="%1.1f%%",wedgeprops={"edgecolor":"white"})
plt.title("percentage of worker view job titlt")
plt.show()
```
